```python
exp = oc["EXPIRY DATE"].unique()
for i in range(len(exp)):
    print(exp[i])
arr = []
arr
```

```python
arr= []
for i in range(len(exp)):
    arr.append(pd.read_csv(f"option-chain-ED-BANKNIFTY-{exp[i]}.csv", header = 1))
```

```python
count = 0
for i in range(len(arr)):
    arr[i] = arr[i].drop(["NO TEXT", "NO TEXT.1"], axis = 1)
    count += len(arr[i])
print(arr[1])
```

```python
print(((arr[1].columns)))
```

```python
for i in range(len(arr)):
    df1 = arr[i][['OI', 'CHNG IN OI', 'VOLUME', 'IV', 'LTP', 'CHNG', 'BID QTY', 'BID', 'ASK', 'ASK QTY', 'STRIKE']]
    df2 = arr[i][['OI.1', 'CHNG IN OI.1', 'VOLUME.1', 'IV.1', 'LTP.1', 'CHNG.1', 'BID QTY.1', 'BID.1', 'ASK.1', 'ASK QTY.1', 'STRIKE']]
    df2.columns = ['OI', 'CHNG IN OI', 'VOLUME', 'IV', 'LTP', 'CHNG', 'BID QTY', 'BID', 'ASK', 'ASK QTY', 'STRIKE']
    df1.loc[ : , "OPTION TYPE"]= "Call"
    df2.loc[ : , "OPTION TYPE"]= "Put"
    arr[i]= pd.concat([df1, df2], ignore_index=True)
    for j in range(len(arr[i])):
        arr[i].loc[j, "EXPIRY DATE"] = exp[i]
```

```python
oca = pd.DataFrame()
for i in range(len(arr)):
    df = pd.concat([oca, arr[i]], ignore_index = True)
    oca = df

oca["EXPIRY DATE"].unique()
```

```python
oca.columns
```

```python
oca = oca[["IV", "LTP", "STRIKE", "OPTION TYPE", "EXPIRY DATE"]]
oca
```

```python
oc.rename(columns = {"STRIKE PRICE": "STRIKE"}, inplace = True)
oc
```

```python
for i in oc.index:
    for j in oca.index:
        if((oc.loc[i, "EXPIRY DATE"] == oca.loc[j, "EXPIRY DATE"]) & (oc.loc[i, "OPTION TYPE"] == oca.loc[j, "OPTION TYPE"]) & (oc.loc[i, "STRIKE"] == oca.loc[j, "STRIKE"])):
            oc.loc[i, "IV"] = oca.loc[j, "IV"]
            break
```
